```python
# Humanクラスの定義
class Human:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def check_adult(self):
        if self.age >= 20:
            print(f"{self.name}は大人です。")
        else:
            print(f"{self.name}は大人ではありません。")


# Humanクラスのインスタンスを複数生成
human_list = [
    Human("太郎", 25),
    Human("花子", 18),
    Human("次郎", 20),
    Human("美咲", 15),
    Human("光",16),
    Human("蒼",24),
    Human("宇",12)
]

# リストの要素数分だけcheck_adultメソッドを呼び出す
for human in human_list:
    human.check_adult()
```
