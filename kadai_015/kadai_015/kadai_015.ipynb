```python
# Humanクラスの定義
class Human:
    # コンストラクタ（属性の定義）
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # nameとageを表示するメソッド
    def printinfo(self):
        print(self.name)
        print(self.age)


# Humanクラスのインスタンスを作成し、変数に代入
human = Human("賢治", 20)

# メソッドを呼び出して表示
human.printinfo()
```
